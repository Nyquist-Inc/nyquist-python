# 02 · Yield curve

Build a US Treasury yield curve from constant-maturity yield indices,
each pulled through `nq.marketdata.history`. The latest close of each
index is one point on the curve (yfinance treasury indices quote in %).

```python
from nyquist import Nyquist

nq = Nyquist()
```

Treasury constant-maturity yield indices (13-week, 5y, 10y, 30y):

```python
TENORS = {"13W": "^IRX", "5Y": "^FVX", "10Y": "^TNX", "30Y": "^TYX"}

points = {}
for label, ticker in TENORS.items():
    hist = nq.marketdata.history(ticker, days=10)
    if not hist.empty:
        points[label] = float(hist["close"].iloc[-1])
points
```

## Plot the term structure

```python
tenors = list(points.keys())
yields = list(points.values())
ax = nq.plot.curve(tenors, yields, title="US Treasury yield curve")
ax
```

> Empty `points` means the Golden Record store has no recent data for
> those indices in this environment — the SDK returns empty frames rather
> than fabricating a curve.
