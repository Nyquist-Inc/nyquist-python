# 04 · Ask the ontology (the semantic layer)

The ontology is Nyquist's typed, bitemporal object graph — VaR reports,
Greeks, regimes, counterparties, instruments, agent verdicts all live
there as first-class objects. Query it in natural language, or by
vector similarity.

```python
from nyquist import Nyquist

nq = Nyquist()
```

## Grounded natural-language answer

`ask` returns an `AskResult` with `.answer` (prose) and `.rows`
(the supporting objects as a DataFrame).

```python
res = nq.ontology.ask(
    "Which counterparty carries our largest exposure?",
    language="en",
)
print(res.answer)
print("intent:", res.intent, "| provider:", res.provider, "| error:", res.error)
```

```python
res.rows
```

## Semantic similarity search

`search` embeds the query and returns the nearest typed objects with a
`distance` column (lower = more similar). Empty when embeddings are
unconfigured — then fall back to `ask`.

```python
hits = nq.ontology.search("AI semiconductor supply chain risk", k=10)
hits
```

```python
# Filter to a single object kind
nq.ontology.search("rising rate volatility", k=5, kind="instrument")
```
