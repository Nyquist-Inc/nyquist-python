# 01 · Quickstart

Your Nyquist research environment in VS Code (BQuant-style).

**Before running:** set `NYQUIST_API_KEY` in your shell (the devcontainer
forwards it from the host). Get a B2B key from the Nyquist portal → API Keys.
Data is pulled **live** from the API — cells below have no baked outputs.

```python
from nyquist import Nyquist

nq = Nyquist()  # api_key defaults to the NYQUIST_API_KEY env var
nq
```

## Pull OHLCV history

`nq.marketdata.history` returns a date-indexed pandas DataFrame (open/high/low/close/volume).

```python
df = nq.marketdata.history("NVDA", days=365)
print(df.attrs.get("source"), "·", len(df), "bars")
df.tail()
```

## Plot the close

`nq.plot` helpers need the `[plot]` extra (`pip install 'nyquist[plot]'`), already installed in the devcontainer.

```python
ax = nq.plot.price(df, column="close", title="NVDA — daily close")
ax
```

## A point quote

`price` (equity) and `fx` (currency pair) return plain dicts.

```python
nq.marketdata.price("NVDA")
```

```python
nq.marketdata.fx("EUR", "USD")
```
